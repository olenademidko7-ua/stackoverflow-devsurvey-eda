Завдання 3. Статистичний аналіз досвіду респондентів

```python
# колонка WorkExp у датасеті з Stack Overflow часто має тип object (рядки), тому її потрібно перетворити в число.
# перетворюємо в числовий формат
df["WorkExp"] = pd.to_numeric(df["WorkExp"], errors="coerce")

print(f'Міри центральної тенденції:')

# середнє
print(f'Mean: {round(df["WorkExp"].mean(), 1)}')

# медіана
print(f'Median: {df["WorkExp"].median()}')

# мода
print(f'Mode: {df["WorkExp"].mode()[0]}')

# пропущені значення були оброблені шляхом перетворення у NaN, і при розрахунку мір центральної тенденції вони автоматично ігноруються методами pandas.
```

Завдання 5. Визначення популярності Python

```python
# створюємо новий стовпець worked_with_python (True/False)
df["worked_with_python"] = df["LanguageHaveWorkedWith"].str.contains(
    "Python", case=False, na=False
)
```

```python
# рахуємо кількість респондентів, які використовують Python
python_users = df["worked_with_python"].sum()
```

```python
# загальна кількість респондентів
total = df.shape[0]
```

```python
# обчислюємо відсоток
python_percent = (python_users / total) * 100

print(f"Відсоток респондентів: {round(python_percent, 1)}%")
```

Завдання 6. Аналіз шляхів навчання програмуванню

```python
# створюємо новий стовпець Learned_with_online_courses (True/False)
df["Learned_with_online_courses"] = df["LearnCode"].str.contains(
     "Online Courses", case=False, na=False
)
```

```python
# підраховуємо кількість респондентів
online_courses_count = df["Learned_with_online_courses"].sum()

int(online_courses_count)
```

Завдання 7. Географічний аналіз компенсації Python-розробників

```python
# cтворюємо новий стовпець — чи працює респондент з Python
df['Worked_with_Python'] = df['LanguageHaveWorkedWith'].str.contains('Python', na=False)
```

```python
# фільтруємо тільки Python-розробників
python_df = df[df['Worked_with_Python'] == True]
```

```python
# прибираємо пропущені значення в компенсації
python_df = python_df.dropna(subset=['ConvertedCompYearly', 'Country'])
```

```python
# групуємо по країнах і рахуємо середнє та медіану
py_progers_by_country = python_df.groupby('Country')['ConvertedCompYearly'].agg(
    average_compensation='mean',
    median_compensation='median'
).reset_index()
py_progers_by_country
```

Завдання 8. Аналіз освіти найбільш оплачуваних спеціалістів

```python
# вибираємо потрібні стовпці 'ResponseId', 'EdLevel', 'ConvertedCompYearly'
# прибираємо рядки з пропущеною зарплатою (dropna)
# сортуємо за зарплатою у спадному порядку (sort_values)
# оновлюємо індекси (reset_index)
# беремо топ-5 (head(5))
top_5_education = (
    df[['ResponseId', 'EdLevel', 'ConvertedCompYearly']]
    .dropna(subset=['ConvertedCompYearly'])
    .sort_values(by='ConvertedCompYearly', ascending=False)
    .reset_index(drop=True)
    .head(5)
)
top_5_education
```

Завдання 9. Аналіз популярності Python по віковим категоріям

```python
# прибираємо рядки без віку або мов програмування
df_copy = df_copy.dropna(subset=['Age', 'LanguageHaveWorkedWith'])
```

```python
# створюємо колонку: чи використовує респондент Python
df_copy['UsesPython'] = df_copy['LanguageHaveWorkedWith'].str.contains('Python', case=False, na=False)
```

```python
# групуємо по віковим категоріям і рахуємо відсоток
python_by_age = (
    df_copy
    .groupby('Age')['UsesPython']
    .mean() * 100
)
```

```python
# перетворюємо у таблицю (скидаємо індекс)
python_by_age = python_by_age.reset_index()
```

```python
# перейменовуємо колонку для зрозумілості
# округлюємо (round)
python_by_age = python_by_age.rename(columns={'UsesPython': 'Python_%'})

python_by_age['Python_%'] = python_by_age['Python_%'].round(2)

python_by_age
```

Завдання 10. Аналіз індустрій серед високооплачуваних віддалених працівників

```python
# обчислюємо 75-й перцентиль компенсації
q75 = df['ConvertedCompYearly'].quantile(0.75)
q75
```

```python
# фільтруємо респондентів за умовами (Знаходяться у 75 перцентилі за компенсацією & Працюють віддалено)
filtered_df = df[
    (df['ConvertedCompYearly'] > q75) &
    (df['RemoteWork'].isin(['Remote']))
]

```

```python
# популярні індустрії
industry_counts = (
    filtered_df['Industry']
    .value_counts()
    .reset_index()
)

industry_counts.columns = ['Industry', 'Count']

industry_counts
```

ВІЗУАЛІЗАЦІЯ РЕЗУЛЬТАТІВ

```python
import matplotlib.pyplot as plt

remote_count = (df["RemoteWork"] == "Remote").sum()
other_count = len(df) - remote_count
total = len(df)

fig, ax = plt.subplots(figsize=(7, 5))
fig.patch.set_facecolor("#1a1a1a")
ax.set_facecolor("#1a1a1a")

sizes  = [remote_count, other_count]
labels = ["Remote", "Інші формати"]
colors = ["#FF6B00", "#444444"]
explode = (0.05, 0)  # виділяємо Remote

wedges, texts, autotexts = ax.pie(
    sizes,
    labels=labels,
    colors=colors,
    explode=explode,
    autopct="%1.1f%%",
    startangle=90,
    textprops={"color": "white", "fontsize": 12}
)

for autotext in autotexts:
    autotext.set_fontweight("bold")
    autotext.set_fontsize(13)

ax.set_title("Віддалена робота розширює\nможливості заробітку",
             fontsize=14, fontweight="bold", pad=20, color="white")

plt.tight_layout()
plt.show()
```

```python
top_languages = df["LanguageHaveWorkedWith"].str.split(";").explode().value_counts().head(10)

fig, ax = plt.subplots(figsize=(9, 5))
fig.patch.set_facecolor("#1a1a1a")
ax.set_facecolor("#1a1a1a")

colors = ["#3776AB" if lang == "Python" else "#FF6B00" for lang in top_languages.index]

ax.barh(top_languages.index[::-1], top_languages.values[::-1], color=colors[::-1])

ax.set_title("Популярність Python у 2025 році\nчерез розвиток AI та Data Science",
             fontsize=14, fontweight="bold", pad=15, color="white")
ax.set_xlabel("Кількість респондентів", color="white")
ax.tick_params(colors="white")
ax.spines[["top", "right"]].set_visible(False)
ax.spines[["left", "bottom"]].set_edgecolor("#444444")
plt.tight_layout()
plt.show()
```

```python
# очищаємо дані
df_salary = df[["WorkExp", "ConvertedCompYearly"]].dropna()
df_salary = df_salary[df_salary["ConvertedCompYearly"] < 300000]

# групуємо по досвіду
bins   = [0, 2, 5, 10, 20, 100]
labels = ["0–2", "3–5", "6–10", "11–20", "20+"]
df_salary["exp_group"] = pd.cut(df_salary["WorkExp"], bins=bins, labels=labels)

salary_by_exp = df_salary.groupby("exp_group", observed=True)["ConvertedCompYearly"].median().reset_index()

fig, ax = plt.subplots(figsize=(9, 5))
fig.patch.set_facecolor("#1a1a1a")
ax.set_facecolor("#1a1a1a")

bars = ax.bar(
    salary_by_exp["exp_group"],
    salary_by_exp["ConvertedCompYearly"],
    color="#FF6B00",
    edgecolor="#1a1a1a",
    width=0.5
)

# підписи над стовпцями
for bar, val in zip(bars, salary_by_exp["ConvertedCompYearly"]):
    ax.text(bar.get_x() + bar.get_width() / 2,
            val + 1000,
            f'${val:,.0f}',
            ha="center", fontsize=10, fontweight="bold", color="white")

ax.set_title("Високі зарплати пов'язані з досвідом\nне лише з освітою",
             fontsize=14, fontweight="bold", pad=15, color="white")
ax.set_xlabel("Роки досвіду", color="white")
ax.set_ylabel("Медіанна зарплата (USD)", color="white")
ax.tick_params(colors="white")
ax.spines[["top", "right"]].set_visible(False)
ax.spines[["left", "bottom"]].set_edgecolor("#444444")
plt.tight_layout()
plt.show()
```
